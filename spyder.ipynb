# Task #2: Loading and Visualizing Images

```python
import cv2

img1= cv2.imread("./images/image.png") # Load the image data into a numpy matrix
img2= cv2.imread("./images/image copy.png") # Load the image data into a numpy matrix

img1rgb= cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2rgb= cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

import matplotlib.pyplot as plt

# This will create the 1x2 subplot grid
fig, ix = plt.subplots(1, 2)

# Plot the first image on the left axis
ix[0].imshow(img1rgb)
ix[0].set_title("Original Image")
ix[0].axis("off") # Hides the x and y axis ticks

# Plot the second image on the right axis
ix[1].imshow(img2rgb)
ix[1].set_title("Image Copy")
ix[1].axis("off")

# Display the figure
plt.show()
```

```python
from PIL import Image

img1pil= Image.open("./images/image.png")
img2pil= Image.open("./images/image copy.png")

import matplotlib.pyplot as plt
import matplotlib.cm as cm

plt.imshow(img1pil, cmap= cm.Greys_r)
plt.show()

plt.imshow(img2pil, cmap= cm.Greys_r)
plt.show()
```

```python
cv2.imwrite("./images/New1img2rgb.png", img2rgb)
```

```python
image_array= np.array(img2pil)
new_vis= cv2.GaussianBlur(image_array, ksize= (0, 0), sigmaX= 3)

plt.imshow(new_vis)
plt.show()

image_saving= cv2.cvtColor(new_vis, cv2.COLOR_RGB2BGR)

cv2.imwrite("./images/Img2pilbblurry.png", new_vis)
cv2.imwrite("./images/Img2pilbblurry2.png", image_saving)
```

```python
image_pil1= Image.fromarray(image_saving)
image_pil2= Image.fromarray(new_vis)

image_pil1.save("./images/Pilimg21.png")
image_pil2.save("./images/Pilimg22.png")
```
